```python
def count_vowels(text):
    vowels = "aeiouAEIO"
    count = 0
    for char in text:
       if char in vowels:
           count +=1
    return count
n = input("enter a string")
result = count_vowels(n)   
print(f"total word :{result}")    
```

```python
def get_factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i   
    return result 
num = int(input("Enter a number: "))
print(f"The factorial of {num} is {get_factorial(num)}")
```
